# Day 018: Null Space and Column Space


## 1. Objective
> To master the two fundamental subspaces of a matrix:  
> * Column Space ($C(\mathbf{A})$): The "Output Realm"—the set of all possible vectors that can be reached by the matrix.  
> * Null Space ($N(\mathbf{A})$): The "Zero-Zone"—the set of all non-zero vectors that the matrix smashes into the origin $(0,0)$.

## 2. Mathematical Foundation
> A. The Column Space ($C(\mathbf{A})$)The Column Space is the span of the columns of $\mathbf{A}$. If $\mathbf{A}$ transforms space, the Column Space is the subset of the destination space that is actually "hit" by the transformation.Rank-Column Connection: The dimension of the Column Space is exactly equal to the Rank of the matrix.   
>  
> B. The Null Space ($N(\mathbf{A})$)The Null Space contains all vectors $\mathbf{x}$ such that:$$\mathbf{A}\mathbf{x} = \mathbf{0}$$These vectors are "hidden" by the matrix because their output is indistinguishable from the origin.    
> 
> C. The Fundamental Theorem of Linear AlgebraFor any matrix $\mathbf{A}$ of shape $(m \times n)$:$$\text{dim}(C(\mathbf{A})) + \text{dim}(N(\mathbf{A})) = n$$(The number of "surviving" dimensions plus the number of "crushed" dimensions must equal the total original dimensions.)

## 3. Real-World & AI Applications
> * **Error Correction (Coding Theory):** In telecommunications, we design matrices where the "noise" (errors) is forced into the Null Space. By checking if a signal lands in the Null Space, we can detect if data was corrupted during transmission.  
> * **AI/ML (Feature Engineering):** When preprocessing data for a Neural Network, if your features contain redundant variables, they contribute to the Null Space. By computing the basis of the Null Space, you can explicitly identify the exact combinations of features that are mathematically useless or redundant and prune them to make the model faster.

## 4. Algorithmic Strategy
> * **Computation:** Use `scipy.linalg.null_space(A)` to find the basis of the Null Space.  
> * **Numerical Warning:** Because of machine epsilon, the Null Space basis is often computed using the **SVD** (Singular Value Decomposition). We look for singular values that are extremely close to zero (e.g., $< 10^{-12}$) and treat those vectors as the Null Space.

## 5. Implementation

```python
import numpy as np
from scipy.linalg import null_space

A = np.array([[1, 2, 3],
              [2, 4, 6]]) # Row 2 is 2x Row 1

print("--- 1. Subspace Analysis ---")
# Column Space Basis
# Note: Use SVD to find orthonormal basis for Col Space
U, S, Vh = np.linalg.svd(A)
col_space_basis = U[:, :np.linalg.matrix_rank(A)]

print("Basis of Column Space:")
print(col_space_basis)

# Null Space Basis
ns = null_space(A)
print("\nBasis of Null Space:")
print(ns)

# Verification: A @ ns should be essentially zero
print("\nVerification (A @ Null_Space_Basis):")
print(np.round(A @ ns, 10))
```

## 6. Verification

```python
print("\n--- 2. Fundamental Theorem Verification ---")
# dim(Col) + dim(Null) = n (columns)
rank = np.linalg.matrix_rank(A)
null_dim = ns.shape[1]
n_cols = A.shape[1]

print(f"Rank (Dim ColSpace): {rank}")
print(f"Dim NullSpace:       {null_dim}")
print(f"Total Columns (n):   {n_cols}")
print(f"Theorem Holds:       {rank + null_dim == n_cols}")
```

## 7. Complexity Analysis
> - **Time Complexity:** $\mathcal{O}(m \cdot n^2 + n^3)$

This cubic scaling is why finding exact subspaces for massive matrices (like deep learning weights) requires specialized, highly optimized linear algebra libraries (like LAPACK).
